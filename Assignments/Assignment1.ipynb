### 1. Addition Rule of Probability

Consider two events A and B such that P(A)=0.4, P(B)=0.5, and P (A∩B)=0.2<br>
(a) Find P (A∪B) using the addition rule of probability.<br>
(b) Are the events A and B mutually exclusive? Justify your answer.<br>

```python
p_A = 0.4
p_B = 0.5
p_A_intersection_B = 0.2

p_A_union_B = p_A + p_B - p_A_intersection_B
print('P(A ∪ B) =', p_A_union_B)
```

```python
if p_A_union_B == 0:
    print('Mutually Exclusive')
else:
    print('Not Mutually Exclusive')
```
