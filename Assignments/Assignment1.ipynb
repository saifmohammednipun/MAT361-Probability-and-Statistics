### 1. Addition Rule of Probability

Consider two events A and B such that P(A)=0.4, P(B)=0.5, and P (A∩B)=0.2<br>
(a) Find P (A∪B) using the addition rule of probability.<br>
(b) Are the events A and B mutually exclusive? Justify your answer.<br>

```python
p_A = 0.4
p_B = 0.5
p_A_intersection_B = 0.2

p_A_union_B = p_A + p_B - p_A_intersection_B
print('P(A ∪ B) =', p_A_union_B)
```

```python
if p_A_union_B == 0:
    print('Mutually Exclusive')
else:
    print('Not Mutually Exclusive')
```

### 2. Conditional Probability

<p>A factory produces electronic devices, 10% of which are defective. A quality
control process is 90% eﬀective in detecting defective items and 95% eﬀective
in correctly identifying non-defective items. Suppose an item is tested, and the
test indicates that it is defective. What is the probability that the item is indeed
defective? Use conditional probability.</p>

```python
p_Defective = 0.10
p_Positive_given_Defective = 0.90
p_Negative_given_NonDefective = 0.95

p_Positive_and_Defective = p_Positive_given_Defective * p_Defective 
p_Positive_and_NonDefective = (1 - p_Negative_given_NonDefective) * (1 - p_Defective)
p_Positive = (p_Positive_and_Defective) + (p_Positive_and_NonDefective)

# Conditional Probability
p_Defective_given_Positive = p_Positive_and_Defective / p_Positive

p_Defective_given_Positive = round(p_Defective_given_Positive, 3)

print('P(Defective|Positive) =', p_Defective_given_Positive)
```

### 3. Bayesian Probability

<p>A certain disease occurs in 1% of the population. A test for the disease has a 99% sensitivity (the probability of testing positive given that one has the
disease) and a 95% specificity (the probability of testing negative given that
one does not have the disease). </p>
(a) If a person tests positive, what is the probability that they actually have the disease? Use Bayes’ Theorem to compute this.</br>
(b) Interpret the result in terms of false positives.

```python
p_Disease = 0.01
p_Positive_given_Disease = 0.99
p_Negative_given_NoDisease = 0.95

p_Positive_and_Disease = p_Positive_given_Disease * p_Disease
p_Positive_and_NoDisease = (1 - p_Negative_given_NoDisease) * (1 - p_Disease)
p_Positive = (p_Positive_and_Disease) + (p_Positive_and_NoDisease)

# Bayes Theorem
p_Disease_given_Positive = (p_Positive_given_Disease * p_Disease) / p_Positive

p_Disease_given_Positive = round(p_Disease_given_Positive, 3)

print('P(Disease|Positive) =', p_Disease_given_Positive)
```

```python
p_NoDisease_given_Positive = 1 - p_Disease_given_Positive

p_NoDisease_given_Positive = round(p_NoDisease_given_Positive, 3)
print('P(No Disease|Positive) =', p_NoDisease_given_Positive)
```
